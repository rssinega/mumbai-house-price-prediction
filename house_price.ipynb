```python
dummies=pd.get_dummies(df['furnished'],dtype=int)
dummies
```

```python
merged=pd.concat([df,dummies],axis="columns")
print(merged)
```

```python
X=pd.DataFrame(merged)
X.head()
```

```python
new_X=X.drop(['furnished','locality','price','total_floors','property_type','title','city'],axis='columns')
new_X
```

```python
new_y=df['price']
new_y
```

```python
from sklearn.model_selection import train_test_split
new_X_train,new_X_test,new_y_train,new_y_test=train_test_split(new_X,new_y,test_size=0.2,random_state=42)
```

```python
# linear regression
```

```python
from sklearn.linear_model import LinearRegression
lines=LinearRegression()
lines.fit(new_X_train,new_y_train)
```

```python
lines1=lines.predict(new_X_test)
```

```python
lines.score(new_X_train,new_y_train)
```

```python
lines.score(new_X_test,new_y_test)
```
